```python
from langchain_huggingface import ChatHuggingFace
from langchain_huggingface import HuggingFaceEndpoint
```

```python
llm = HuggingFaceEndpoint(
    repo_id="deepseek-ai/DeepSeek-R1",
    task="text-generation",
    max_new_tokens=512,
    # do_sample=False,
    # repetition_penalty=1.03,
)
```

```python
chat_model = ChatHuggingFace(llm=llm)
```

```python
chat_model.invoke("hello how are you?")
```

```python
from google.colab import userdata
import os
from langchain_google_genai import ChatGoogleGenerativeAI

# Load Gemini API key from Colab Secrets
os.environ["GOOGLE_API_KEY"] = userdata.get("GEMINI_API_KEY")

# Create Gemini chat model
chat_model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash-lite",
    temperature=0.7,
    max_output_tokens=512
)

# Test invoke
response = chat_model.invoke("hello how are you?")

print(response.content)
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

template="""Question: {question}
give me a answer in detail manner and in step by step manner"""

prompt=PromptTemplate(template=template,input_variables=["question"])

llm_chain = (
        {"question": RunnablePassthrough()}
        | prompt
        | chat_model
        | StrOutputParser()
    )

question="who is a first president of INDIA?"

llm_chain.invoke({"question":question})
```
